```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Placeholder paths for datasets
bike_dataset_path = 'path/to/bike_dataset.csv'
car_sales_dataset_path = 'path/to/car_sales_dataset.csv'
amazon_sales_dataset_path = 'path/to/amazon_sales_dataset.csv'
spotify_data_path = 'path/to/spotify_data.csv'

###############################
# Bike Details Dataset (EDA-1)
###############################

# Load dataset
bike_data = pd.read_csv(bike_dataset_path)

# Basic Info
bike_data.info()

# 1. Range of selling prices
selling_price_range = (bike_data['selling_price'].min(), bike_data['selling_price'].max())
print(f"Range of selling prices: {selling_price_range}")

# 2. Median selling price
median_selling_price = bike_data['selling_price'].median()
print(f"Median selling price: {median_selling_price}")

# 3. Most common seller type
most_common_seller_type = bike_data['seller_type'].mode()[0]
print(f"Most common seller type: {most_common_seller_type}")

# 4. Bikes driven more than 50,000 km
bikes_above_50000_km = bike_data[bike_data['km_driven'] > 50000].shape[0]
print(f"Number of bikes driven more than 50,000 km: {bikes_above_50000_km}")

# 5. Average km_driven for each ownership type
avg_km_per_owner = bike_data.groupby('owner')['km_driven'].mean()
print("Average km_driven for each ownership type:\n", avg_km_per_owner)

# 6. Proportion of bikes from 2015 or older
year_condition = bike_data['year'] <= 2015
proportion_older_bikes = year_condition.mean()
print(f"Proportion of bikes from 2015 or older: {proportion_older_bikes:.2%}")

# 7. Missing values
missing_values = bike_data.isnull().sum()
print("Missing values:\n", missing_values)

# 8. Highest ex_showroom_price and bike details
max_ex_showroom_price = bike_data['ex_showroom_price'].max()
highest_priced_bike = bike_data[bike_data['ex_showroom_price'] == max_ex_showroom_price]
print("Highest ex_showroom_price:", max_ex_showroom_price)
print("Details of the bike:", highest_priced_bike)

# 9. Total number of bikes listed by each seller type
total_bikes_by_seller = bike_data['seller_type'].value_counts()
print("Total bikes by seller type:\n", total_bikes_by_seller)

# 10. Relationship between selling_price and km_driven for first-owner bikes
first_owner_bikes = bike_data[bike_data['owner'] == '1st owner']
plt.figure(figsize=(8, 5))
sns.scatterplot(x='km_driven', y='selling_price', data=first_owner_bikes)
plt.title('Relationship between Selling Price and KM Driven (1st Owner Bikes)')
plt.xlabel('KM Driven')
plt.ylabel('Selling Price')
plt.show()

# 11. Identify and remove outliers in the km_driven column using the IQR method
Q1 = bike_data['km_driven'].quantile(0.25)
Q3 = bike_data['km_driven'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
bike_data_no_outliers = bike_data[(bike_data['km_driven'] >= lower_bound) & (bike_data['km_driven'] <= upper_bound)]
print("Removed outliers from km_driven column.")

# 12. Bivariate analysis: year vs. selling_price
plt.figure(figsize=(10, 6))
sns.boxplot(x='year', y='selling_price', data=bike_data_no_outliers)
plt.title('Year vs. Selling Price')
plt.xticks(rotation=45)
plt.show()

# 13. Average depreciation in selling price based on bike's age
current_year = 2024
bike_data['bike_age'] = current_year - bike_data['year']
avg_depreciation = bike_data.groupby('bike_age')['selling_price'].mean()
print("Average depreciation in selling price by bike age:\n", avg_depreciation)

# 14. Bikes priced significantly above average for their manufacturing year
avg_price_by_year = bike_data.groupby('year')['selling_price'].mean()
above_avg_bikes = bike_data[bike_data['selling_price'] > bike_data['year'].map(avg_price_by_year)]
print("Bikes priced significantly above average for their year:\n", above_avg_bikes[['name', 'year', 'selling_price']])

# 15. Correlation matrix and heatmap for numeric columns
corr_matrix = bike_data_no_outliers.corr()
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()

###############################
# Car Sales Dataset (EDA-2)
###############################

# Load dataset
car_data = pd.read_csv(car_sales_dataset_path)

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Placeholder paths for datasets
bike_dataset_path = 'path/to/bike_dataset.csv'
car_sales_dataset_path = 'path/to/car_sales_dataset.csv'
amazon_sales_dataset_path = 'path/to/amazon_sales_dataset.csv'
spotify_data_path = 'path/to/spotify_data.csv'

###############################
# Bike Details Dataset (EDA-1)
###############################

# Load dataset
bike_data = pd.read_csv(bike_dataset_path)

# Basic Info
bike_data.info()

# 1. Range of selling prices
selling_price_range = (bike_data['selling_price'].min(), bike_data['selling_price'].max())
print(f"Range of selling prices: {selling_price_range}")

# 2. Median selling price
median_selling_price = bike_data['selling_price'].median()
print(f"Median selling price: {median_selling_price}")

# 3. Most common seller type
most_common_seller_type = bike_data['seller_type'].mode()[0]
print(f"Most common seller type: {most_common_seller_type}")

# 4. Bikes driven more than 50,000 km
bikes_above_50000_km = bike_data[bike_data['km_driven'] > 50000].shape[0]
print(f"Number of bikes driven more than 50,000 km: {bikes_above_50000_km}")

# 5. Average km_driven for each ownership type
avg_km_per_owner = bike_data.groupby('owner')['km_driven'].mean()
print("Average km_driven for each ownership type:\n", avg_km_per_owner)

# 6. Proportion of bikes from 2015 or older
year_condition = bike_data['year'] <= 2015
proportion_older_bikes = year_condition.mean()
print(f"Proportion of bikes from 2015 or older: {proportion_older_bikes:.2%}")

# 7. Missing values
missing_values = bike_data.isnull().sum()
print("Missing values:\n", missing_values)

# 8. Highest ex_showroom_price and bike details
max_ex_showroom_price = bike_data['ex_showroom_price'].max()
highest_priced_bike = bike_data[bike_data['ex_showroom_price'] == max_ex_showroom_price]
print("Highest ex_showroom_price:", max_ex_showroom_price)
print("Details of the bike:", highest_priced_bike)

# 9. Total number of bikes listed by each seller type
total_bikes_by_seller = bike_data['seller_type'].value_counts()
print("Total bikes by seller type:\n", total_bikes_by_seller)

# 10. Relationship between selling_price and km_driven for first-owner bikes
first_owner_bikes = bike_data[bike_data['owner'] == '1st owner']
plt.figure(figsize=(8, 5))
sns.scatterplot(x='km_driven', y='selling_price', data=first_owner_bikes)
plt.title('Relationship between Selling Price and KM Driven (1st Owner Bikes)')
plt.xlabel('KM Driven')
plt.ylabel('Selling Price')
plt.show()

# 11. Identify and remove outliers in the km_driven column using the IQR method
Q1 = bike_data['km_driven'].quantile(0.25)
Q3 = bike_data['km_driven'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
bike_data_no_outliers = bike_data[(bike_data['km_driven'] >= lower_bound) & (bike_data['km_driven'] <= upper_bound)]
print("Removed outliers from km_driven column.")

# 12. Bivariate analysis: year vs. selling_price
plt.figure(figsize=(10, 6))
sns.boxplot(x='year', y='selling_price', data=bike_data_no_outliers)
plt.title('Year vs. Selling Price')
plt.xticks(rotation=45)
plt.show()

# 13. Average depreciation in selling price based on bike's age
current_year = 2024
bike_data['bike_age'] = current_year - bike_data['year']
avg_depreciation = bike_data.groupby('bike_age')['selling_price'].mean()
print("Average depreciation in selling price by bike age:\n", avg_depreciation)

# 14. Bikes priced significantly above average for their manufacturing year
avg_price_by_year = bike_data.groupby('year')['selling_price'].mean()
above_avg_bikes = bike_data[bike_data['selling_price'] > bike_data['year'].map(avg_price_by_year)]
print("Bikes priced significantly above average for their year:\n", above_avg_bikes[['name', 'year', 'selling_price']])

# 15. Correlation matrix and heatmap for numeric columns
corr_matrix = bike_data_no_outliers.corr()
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()

###############################
# Car Sales Dataset (EDA-2)
###############################

# Load dataset
car_data = pd.read_csv(car_sales_dataset_path)

# Questions 1 to 10 (already included above)

# 11. Seasonal trends in car sales based on the date of sale
car_data['Date'] = pd.to_datetime(car_data['Date'])
car_data['Month'] = car_data['Date'].dt.month
monthly_sales = car_data.groupby('Month').size()
plt.figure(figsize=(10, 5))
plt.plot(monthly_sales.index, monthly_sales.values, marker='o')
plt.title('Seasonal Trends in Car Sales')
plt.xlabel('Month')
plt.ylabel('Number of Cars Sold')
plt.xticks(range(1, 13))
plt.show()

# 12. Correlation between price, engine size, and annual income
correlation_matrix = car_data[['Price ($)', 'Engine', 'Annual Income']].corr()
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()

# 13. Price distribution across body style and transmission type
plt.figure(figsize=(12, 6))
sns.boxplot(x='Body Style', y='Price ($)', hue='Transmission', data=car_data)
plt.title('Price Distribution by Body Style and Transmission')
plt.xticks(rotation=45)
plt.show()

# 14. Average price variation across models and engine types
avg_price_model_engine = car_data.groupby(['Model', 'Engine'])['Price ($)'].mean()
print("Average price by model and engine type:\n", avg_price_model_engine)

# 15. Top-selling regions
top_regions = car_data['Dealer_Region'].value_counts().head()
print("Top-selling regions:\n", top_regions)

###############################
# Amazon Sales Dataset (EDA-3)
###############################

# Load dataset
amazon_data = pd.read_csv(amazon_sales_dataset_path)

# 1. Average rating for each product category
avg_rating_by_category = amazon_data.groupby('category')['rating'].mean()
print("Average rating by category:\n", avg_rating_by_category)

# 2. Top products by rating count in each category
top_products_by_rating_count = amazon_data.sort_values('rating_count', ascending=False).groupby('category').head(1)
print("Top products by rating count in each category:\n", top_products_by_rating_count[['product_name', 'category', 'rating_count']])

# 3. Distribution of discounted prices vs. actual prices
plt.figure(figsize=(8, 5))
sns.scatterplot(x='actual_price', y='discounted_price', data=amazon_data)
plt.title('Discounted Prices vs. Actual Prices')
plt.xlabel('Actual Price')
plt.ylabel('Discounted Price')
plt.show()

# 4. Average discount percentage across categories
avg_discount_by_category = amazon_data.groupby('category')['discount_percentage'].mean()
print("Average discount percentage by category:\n", avg_discount_by_category)

# 5. Most popular product names
most_popular_products = amazon_data['product_name'].value_counts().head(5)
print("Most popular products:\n", most_popular_products)

# 6. Most popular product keywords
popular_keywords = amazon_data['about_product'].str.split().explode().value_counts().head(10)
print("Most popular product keywords:\n", popular_keywords)

# 7. Correlation between discounted price and rating
corr_discount_rating = amazon_data[['discounted_price', 'rating']].corr().iloc[0, 1]
print(f"Correlation between discounted price and rating: {corr_discount_rating}")

# 8. Top 5 categories by highest ratings
top_categories_by_rating = avg_rating_by_category.sort_values(ascending=False).head(5)
print("Top 5 categories by highest ratings:\n", top_categories_by_rating)

# 9. Distribution of rating counts for most reviewed products
most_reviewed_products = amazon_data.nlargest(5, 'rating_count')
plt.figure(figsize=(8, 5))
sns.barplot(x='product_name', y='rating_count', data=most_reviewed_products)
plt.title('Top 5 Most Reviewed Products')
plt.xticks(rotation=45)
plt.show()

# 10. Recommendations for improvement based on data
print("Recommendations:")
print("1. Optimize pricing strategies in categories with low average ratings.")
print("2. Enhance product descriptions to align with popular keywords.")
print("3. Focus on promoting highly-rated categories and products.")

###############################
# Spotify Data (EDA-4)
###############################

# Load dataset
spotify_data = pd.read_csv(spotify_data_path)

# 1. Null values and duplicates
print("Null values:\n", spotify_data.isnull().sum())
spotify_data = spotify_data.drop_duplicates()
print("Data after removing duplicates:\n", spotify_data.shape)

# 2. Distribution of track popularity
plt.figure(figsize=(8, 5))
sns.histplot(spotify_data['Popularity'], kde=True, bins=20)
plt.title('Distribution of Track Popularity')
plt.xlabel('Popularity')
plt.ylabel('Frequency')
plt.show()

# 3. Relationship between popularity and duration
plt.figure(figsize=(8, 5))
sns.scatterplot(x='Duration (ms)', y='Popularity', data=spotify_data)
plt.title('Popularity vs. Duration')
plt.xlabel('Duration (ms)')
plt.ylabel('Popularity')
plt.show()

# 4. Artist with highest number of tracks
top_artist = spotify_data['Artist'].value_counts().idxmax()
print(f"Artist with highest number of tracks: {top_artist}")

# 5. Top 5 least popular tracks
least_popular_tracks = spotify_data.nsmallest(5, 'Popularity')[['Artist', 'Track Name', 'Popularity']]
print("Top 5 least popular tracks:\n", least_popular_tracks)

# 6. Top 5 most popular artists by average popularity
top_5_artists = spotify_data.groupby('Artist')['Popularity'].mean().nlargest(5)
print("Top 5 most popular artists by average popularity:\n", top_5_artists)

# 7. Most popular tracks of top 5 artists
for artist in top_5_artists.index:
    top_track = spotify_data[spotify_data['Artist'] == artist].nlargest(1, 'Popularity')
    print(f"Most popular track by {artist}: {top_track['Track Name'].values[0]}")

# 8. Pair plot for numerical variables
sns.pairplot(spotify_data[['Popularity', 'Duration (ms)']])
plt.show()

# 9. Variation of track duration across artists
plt.figure(figsize=(12, 6))
sns.boxplot(x='Artist', y='Duration (ms)', data=spotify_data)
plt.title('Track Duration by Artist')
plt.xticks(rotation=45)
plt.show()

# 10. Popularity distribution by artist
plt.figure(figsize=(12, 6))
sns.violinplot(x='Artist', y='Popularity', data=spotify_data)
plt.title('Popularity Distribution by Artist')
plt.xticks(rotation=45)
plt.show()


# 11. Seasonal trends in car sales based on the date of sale
car_data['Date'] = pd.to_datetime(car_data['Date'])
car_data['Month'] = car_data['Date'].dt.month
monthly_sales = car_data.groupby('Month').size()
plt.figure(figsize=(10, 5))
plt.plot(monthly_sales.index, monthly_sales.values, marker='o')
plt.title('Seasonal Trends in Car Sales')
plt.xlabel('Month')
plt.ylabel('Number of Cars Sold')
plt.xticks(range(1, 13))
plt.show()

# 12. Correlation between price, engine size, and annual income
correlation_matrix = car_data[['Price ($)', 'Engine', 'Annual Income']].corr()
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()

# 13. Price distribution across body style and transmission type
plt.figure(figsize=(12, 6))
sns.boxplot(x='Body Style', y='Price ($)', hue='Transmission', data=car_data)
plt.title('Price Distribution by Body Style and Transmission')
plt.xticks(rotation=45)
plt.show()

# 14. Average price variation across models and engine types
avg_price_model_engine = car_data.groupby(['Model', 'Engine'])['Price ($)'].mean()
print("Average price by model and engine type:\n", avg_price_model_engine)

# 15. Top-selling regions
top_regions = car_data['Dealer_Region'].value_counts().head()
print("Top-selling regions:\n", top_regions)

###############################
# Amazon Sales Dataset (EDA-3)
###############################

# Load dataset
amazon_data = pd.read_csv(amazon_sales_dataset_path)

# 1. Average rating for each product category
avg_rating_by_category = amazon_data.groupby('category')['rating'].mean()
print("Average rating by category:\n", avg_rating_by_category)

# 2. Top products by rating count in each category
top_products_by_rating_count = amazon_data.sort_values('rating_count', ascending=False).groupby('category').head(1)
print("Top products by rating count in each category:\n", top_products_by_rating_count[['product_name', 'category', 'rating_count']])

# 3. Distribution of discounted prices vs. actual prices
plt.figure(figsize=(8, 5))
sns.scatterplot(x='actual_price', y='discounted_price', data=amazon_data)
plt.title('Discounted Prices vs. Actual Prices')
plt.xlabel('Actual Price')
plt.ylabel('Discounted Price')
plt.show()

# 4. Average discount percentage across categories
avg_discount_by_category = amazon_data.groupby('category')['discount_percentage'].mean()
print("Average discount percentage by category:\n", avg_discount_by_category)

# 5. Most popular product names
most_popular_products = amazon_data['product_name'].value_counts().head(5)
print("Most popular products:\n", most_popular_products)

# 6. Most popular product keywords
popular_keywords = amazon_data['about_product'].str.split().explode().value_counts().head(10)
print("Most popular product keywords:\n", popular_keywords)

# 7. Correlation between discounted price and rating
corr_discount_rating = amazon_data[['discounted_price', 'rating']].corr().iloc[0, 1]
print(f"Correlation between discounted price and rating: {corr_discount_rating}")

# 8. Top 5 categories by highest ratings
top_categories_by_rating = avg_rating_by_category.sort_values(ascending=False).head(5)
print("Top 5 categories by highest ratings:\n", top_categories_by_rating)

# 9. Distribution of rating counts for most reviewed products
most_reviewed_products = amazon_data.nlargest(5, 'rating_count')
plt.figure(figsize=(8, 5))
sns.barplot(x='product_name', y='rating_count', data=most_reviewed_products)
plt.title('Top 5 Most Reviewed Products')
plt.xticks(rotation=45)
plt.show()

# 10. Recommendations for improvement based on data
print("Recommendations:")
print("1. Optimize pricing strategies in categories with low average ratings.")
print("2. Enhance product descriptions to align with popular keywords.")
print("3. Focus on promoting highly-rated categories and products.")

###############################
# Spotify Data (EDA-4)
###############################

# Load dataset
spotify_data = pd.read_csv(spotify_data_path)

# 1. Null values and duplicates
print("Null values:\n", spotify_data.isnull().sum())
spotify_data = spotify_data.drop_duplicates()
print("Data after removing duplicates:\n", spotify_data.shape)

# 2. Distribution of track popularity
plt.figure(figsize=(8, 5))
sns.histplot(spotify_data['Popularity'], kde=True, bins=20)
plt.title('Distribution of Track Popularity')
plt.xlabel('Popularity')
plt.ylabel('Frequency')
plt.show()

# 3. Relationship between popularity and duration
plt.figure(figsize=(8, 5))
sns.scatterplot(x='Duration (ms)', y='Popularity', data=spotify_data)
plt.title('Popularity vs. Duration')
plt.xlabel('Duration (ms)')
plt.ylabel('Popularity')
plt.show()

# 4. Artist with highest number of tracks
top_artist = spotify_data['Artist'].value_counts().idxmax()
print(f"Artist with highest number of tracks: {top_artist}")

# 5. Top 5 least popular tracks
least_popular_tracks = spotify_data.nsmallest(5, 'Popularity')[['Artist', 'Track Name', 'Popularity']]
print("Top 5 least popular tracks:\n", least_popular_tracks)

# 6. Top 5 most popular artists by average popularity
top_5_artists = spotify_data.groupby('Artist')['Popularity'].mean().nlargest(5)
print("Top 5 most popular artists by average popularity:\n", top_5_artists)

# 7. Most popular tracks of top 5 artists
for artist in top_5_artists.index:
    top_track = spotify_data[spotify_data['Artist'] == artist].nlargest(1, 'Popularity')
    print(f"Most popular track by {artist}: {top_track['Track Name'].values[0]}")

# 8. Pair plot for numerical variables
sns.pairplot(spotify_data[['Popularity', 'Duration (ms)']])
plt.show()

# 9. Variation of track duration across artists
plt.figure(figsize=(12, 6))
sns.boxplot(x='Artist', y='Duration (ms)', data=spotify_data)
plt.title('Track Duration by Artist')
plt.xticks(rotation=45)
plt.show()

# 10. Popularity distribution by artist
plt.figure(figsize=(12, 6))
sns.violinplot(x='Artist', y='Popularity', data=spotify_data)
plt.title('Popularity Distribution by Artist')
plt.xticks(rotation=45)
plt.show()
```
